Crea un nuevo DataFrame con datos de 1 estaciones SENAMHI de diferentes regiones del Perú (https://www.senamhi.gob.pe/site/descarga-datos/).

Los datos de las columnas corresponden a:
- Columna A: Año
- Columna B: Mes
- Columna C: Día
- Columna D: Precipitación acumulada
- Columna E: Temperatura máxima
- Columna F: Temperatura mínima

En las celdas donde aparece -99.9 significa que no hay información disponible para esa variable

Luego:
1. Parsear el nombres de la estación Chosica y leer el archivo.
1. Calcula el promedio mensual de precipitación
1. Calcula el promedio anual de precipitación

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
# Tu código aquí – Ejercicio: Estación

# Leer los datos de las estaciones y crear un DataFrame combinado
import os
files = os.listdir('/content/drive/MyDrive/SENAMHI')  # Lista los archivos en el directorio
files
```

```python
# Parsear el nombre del archivo para obtener el código y nombre de la estación
station = files[3]

codigo = station.replace('Copia de ', '').split('_')[0]
nombre = station.replace('Copia de ', '').split('_')[1].replace('.txt', '')

print(f"Código: {codigo}, Nombre: {nombre}")
```

```python
# Leer el archivo CSV de la estación seleccionada
import pandas as pd
station_df = pd.read_table(f'/content/drive/MyDrive/SENAMHI/{station}', header=None, sep=' ', na_values=['-999','-99.9',"-3256"]) # Sin encabezados
# Formatear las columnas
station_df.columns = ['Año', 'Mes', 'Dia', 'Pr', 'Tmax', 'Tmin'] # Asignar nombres de columnas

station_df
```

```python
# Precipitacion mensual
precipitacion_mensual = station_df.groupby('Mes')['Pr'].mean()

# Precipitacion total anual
precipitacion_anual = station_df.groupby('Año')['Pr'].sum()

# Dia mas lluvioso
dia_mas_lluvioso = station_df.loc[station_df['Pr'].idxmax()]


print(f"Precipitación mensual:\n{precipitacion_mensual}\n")
print(f"Precipitación anual:\n{precipitacion_anual}\n")
print(f"El día más lluvioso fue el {dia_mas_lluvioso['Año']:.0f}-{dia_mas_lluvioso['Mes']:.0f}-{dia_mas_lluvioso['Dia']:.0f} con {dia_mas_lluvioso['Pr']} mm de precipitación.")
```

```python
# Exportar los resultados formateados en CSV
precipitacion_mensual.to_csv('precipitacion_mensual_chosica.csv')
precipitacion_anual.to_csv('precipitacion_anual_chosica.csv')
dia_mas_lluvioso.to_frame().to_csv('dia_mas_lluvioso_chosica.csv')

print("Resultados exportados en CSV")
```

### Ejercicio 2: Datos climáticos multiestación

Crea un nuevo DataFrame con datos de 5 estaciones SENAMHI de diferentes regiones del Perú (https://www.senamhi.gob.pe/site/descarga-datos/).

Luego:
1. Paresear el nombres de la estaciones y leer los archivo.
1. Unir las db (Fecha, Est1, Est2, ...)
1. Calcula el promedio mensual de precipitación por región
2. Identifica el mes más lluvioso en cada estación
4. Exporta el resultado formateado a CSV

```python
# Parsear los nombres de las estaciones
station1 = files[0]
station2 = files[1]
station3 = files[2]
station4 = files[3]
station5 = files[4]
```

```python
# Leer los archivos
df1 = pd.read_table(f'/content/drive/MyDrive/SENAMHI/{station1}', header=None, sep=' ', na_values=['-999','-99.9','-3256'])
df2 = pd.read_table(f'/content/drive/MyDrive/SENAMHI/{station2}', header=None, sep=' ', na_values=['-999','-99.9','-3256'])
df3 = pd.read_table(f'/content/drive/MyDrive/SENAMHI/{station3}', header=None, sep=' ', na_values=['-999','-99.9','-3256'])
df4 = pd.read_table(f'/content/drive/MyDrive/SENAMHI/{station4}', header=None, sep=' ', na_values=['-999','-99.9','-3256'])
df5 = pd.read_table(f'/content/drive/MyDrive/SENAMHI/{station5}', header=None, sep=' ', na_values=['-999','-99.9','-3256'])
```

```python
# Asignar nombres de columnas
df1.columns = ['Año','Mes','Dia','Pr','Tmax','Tmin']
df2.columns = ['Año','Mes','Dia','Pr','Tmax','Tmin']
df3.columns = ['Año','Mes','Dia','Pr','Tmax','Tmin']
df4.columns = ['Año','Mes','Dia','Pr','Tmax','Tmin']
df5.columns = ['Año','Mes','Dia','Pr','Tmax','Tmin']
```

```python
# Crear nuevo DataFrame
df_multi = pd.DataFrame({ 'Est1': df1['Pr'], 'Est2': df2['Pr'], 'Est3': df3['Pr'], 'Est4': df4['Pr'], 'Est5': df5['Pr']})

df_multi
```

```python
# Crear la fecha
df1['Fecha'] = df1['Año'].astype(str) + '-' + df1['Mes'].astype(str) + '-' + df1['Dia'].astype(str)
df2['Fecha'] = df2['Año'].astype(str) + '-' + df2['Mes'].astype(str) + '-' + df2['Dia'].astype(str)
df3['Fecha'] = df3['Año'].astype(str) + '-' + df3['Mes'].astype(str) + '-' + df3['Dia'].astype(str)
df4['Fecha'] = df4['Año'].astype(str) + '-' + df4['Mes'].astype(str) + '-' + df4['Dia'].astype(str)
df5['Fecha'] = df5['Año'].astype(str) + '-' + df5['Mes'].astype(str) + '-' + df5['Dia'].astype(str)
```

```python
# Poner Fecha como índice
df1 = df1.set_index('Fecha')
df2 = df2.set_index('Fecha')
df3 = df3.set_index('Fecha')
df4 = df4.set_index('Fecha')
df5 = df5.set_index('Fecha')

```

```python
# Crear nuevo DataFrame
df_multi = pd.DataFrame({'Est1': df1['Pr'], 'Est2': df2['Pr'],   'Est3': df3['Pr'],  'Est4': df4['Pr'],  'Est5': df5['Pr']})

df_multi.index.name = 'Fecha'

df_multi
```

```python
# Promedio mensual de precipitación por estación
promedio1 = df1.groupby('Mes')['Pr'].mean()
promedio2 = df2.groupby('Mes')['Pr'].mean()
promedio3 = df3.groupby('Mes')['Pr'].mean()
promedio4 = df4.groupby('Mes')['Pr'].mean()
promedio5 = df5.groupby('Mes')['Pr'].mean()

print(f"Est1:\n{promedio1}\n")
print(f"Est2:\n{promedio2}\n")
print(f"Est3:\n{promedio3}\n")
print(f"Est4:\n{promedio4}\n")
print(f"Est5:\n{promedio5}\n")
```

```python
# Identificar el mes más lluvioso en cada estación
mes1 = promedio1.idxmax()
mes2 = promedio2.idxmax()
mes3 = promedio3.idxmax()
mes4 = promedio4.idxmax()
mes5 = promedio5.idxmax()

print(f"Est1: {mes1}")
print(f"Est2: {mes2}")
print(f"Est3: {mes3}")
print(f"Est4: {mes4}")
print(f"Est5: {mes5}")
```

```python
# Crear DataFrame con los promedios mensuales
resultado = pd.DataFrame({ 'Est1': promedio1, 'Est2': promedio2, 'Est3': promedio3,  'Est4': promedio4,  'Est5': promedio5})

resultado.index.name = 'Mes'
```

```python
# Exportar resultados a CSV
df_multi.to_csv('datos_multiestacion.csv')
resultado.to_csv('promedio_mensual_multiestacion.csv')

print("Resultados exportados a CSV")
```
